# Q1) The Probability of Shooting at a Target

A point is chosen at random from a disk of radius $10$. Let $A$ be the event that the point lies within $1$ unit of the boundary.

1. Model the experiment assuming the point is uniformly distributed with respect to area, and compute $P(A)$.
2. Model the experiment assuming the distance to the point from the center is chosen uniformly from $[0,10]$, and the direction is chosen independently and uniformly from $[0,2\pi)$ and compute $P(A)$.
3. Explain why the two answers differ.


Q1) The Probability of Shooting at a Target  
A point is chosen at random from a disk of radius $10$. Let $A$ be the event that the point lies within $1$ unit of the boundary.
The boundary of the disk is at a radius of $10$. Lying within $1$ unit of the boundary means the point's distance from the center, let's call it $r$, is between $9$ and $10$ ($9 \le r \le 10$).
1. Model assuming uniform distribution with respect to area  
Total Area: The total area of the disk with radius $R = 10$ is:
$\text{Area}_{\text{total}} = \pi \times R^2 = \pi \times 10^2 = 100\pi$  
Area of Event A: The region where the point is within $1$ unit of the boundary is an annulus (a ring). Its area is the total area minus the area of the inner disk (radius $r = 9$):
$\text{Area}_A = \pi \times (10^2) - \pi \times (9^2) = 100\pi - 81\pi = 19\pi$  
Probability P(A): Since the point is uniformly distributed by area, the probability is the ratio of the areas:
$P(A) = \frac{\text{Area}_A}{\text{Area}_{\text{total}}} = \frac{19\pi}{100\pi} = \frac{19}{100} = \mathbf{0.19}$
2. Model assuming distance is uniformly distributed from [0, 10]  
In this model, the distance from the center ($r$) is chosen uniformly from the interval $[0, 10]$.  
The direction (angle) is independent and does not affect how far the point is from the center, so we only need to look at the distribution of $r$.  
Probability P(A): Event $A$ occurs when $r$ is in the interval $[9, 10]$. For a uniform 1D distribution, the probability is the length of the target interval divided by the total length of the sample space:
$P(A) = P(9 \le r \le 10) = \frac{10 - 9}{10 - 0} = \frac{1}{10} = \mathbf{0.10}$  
Why the two answers differ

The discrepancy arises because the instruction to "choose a point at random" is mathematically ambiguous. It doesn't clearly define the underlying probability distribution we should use.  
Area-based Uniformity (Model 1): The first approach assumes the probability of a point landing in a specific region is strictly proportional to the area of that region. Because a circle's area expands outward, the outer edges of the disk contain vastly more physical space than the center. Therefore, a randomly placed point has a much higher natural chance of landing near the boundary, leading to $P(A) = \frac{19}{100}$.  
Radius-based Uniformity (Model 2): The second approach treats the radius as a 1D line and distributes the probability evenly along its length. In this model, any 1-unit segment along the radius has an equal chance of being selected, completely ignoring how much 2D area that segment actually sweeps out. Since the target boundary is just a 1-unit interval [9, 10] out of a total 10-unit radial line, the probability is simply $\frac{1}{10}$.  
Ultimately, the paradox happens because we are applying two fundamentally different definitions of "randomness" (2D area vs. 1D distance) to evaluate the exact same geometric event.

# Q2) Bertrand's Paradox

A chord of a circle is chosen “at random.” What is the probability that its length exceeds the side length of the inscribed equilateral triangle?

Show that different mathematically natural models of “random chord” lead to different answers.

Bertrand's Paradox is a classic problem in probability theory that demonstrates how the phrase "chosen at random" is dangerously ambiguous when dealing with infinite sets (like points on a circle).

Because there is no single, universally agreed-upon way to generate a "random chord," we can use three mathematically natural models, all of which give entirely different, equally valid probabilities.

Here are the three standard models to solve this problem:  
Method 1: Random Endpoints (The "Random Angles" Model)  
Model: Choose two random points independently and uniformly on the circumference of the circle, and draw a chord between them.  
Logic: Without loss of generality, we can imagine fixing the first point at one vertex of the inscribed equilateral triangle. This triangle perfectly divides the circle's circumference into three equal arcs.  
Condition: For the randomly drawn chord to be strictly longer than the side of the triangle, the second random point must fall on the arc exactly opposite the first point.  
Probability Calculation: Since the second point is uniformly distributed along the circumference, the probability is simply the length of the favorable arc divided by the total circumference.  
$P = \frac{1}{3}$  
Method 2: Random Radius (The "Random Distance" Model)  
Model: Choose a random radius of the circle, and then choose a random point uniformly along that radius. Draw the chord perpendicular to the radius at that point.  
Logic: Imagine the inscribed equilateral triangle oriented so that one of its sides is perpendicular to our chosen radius. Basic geometry tells us that this side bisects the radius (it crosses the radius exactly halfway between the center and the circumference, at $R/2$).  
Condition: For a chord to be longer than the triangle's side, its midpoint must be closer to the center of the circle than the side of the triangle is.  
Probability Calculation: The midpoint is chosen uniformly along the radius of length $R$. The favorable interval is from the center ($0$) to halfway out ($R/2$).$P = \frac{R/2}{R} = \frac{1}{2}$  
Method 3: Random Midpoint (The "Random Area" Model)  
Model: Choose a random point uniformly distributed over the entire 2D area of the disk. Construct the unique chord that has this point as its midpoint.  
Logic: The large inscribed equilateral triangle actually contains its own smaller inscribed circle (an incircle). The radius of this incircle is exactly half the radius of the larger circle ($R/2$).  
Condition: A chord will be longer than the side of the equilateral triangle if and only if its randomly chosen midpoint falls inside this smaller incircle.  
Probability Calculation: Since the point is distributed uniformly by area, the probability is the ratio of the smaller incircle's area to the larger outer circle's area.  
$\text{Area}_{\text{total}} = \pi R^2$  
$\text{Area}_{\text{favorable}} = \pi \left(\frac{R}{2}\right)^2 = \frac{\pi R^2}{4}$  
$P = \frac{\frac{\pi R^2}{4}}{\pi R^2} = \frac{1}{4}$

#Q8) Probability of Being a Girl?

A family has two children. Assume each child is equally likely to be a boy or a girl, independently of the other.

What is the probability that both children are girls, given that **at least one** of them is a girl?

Q8) Probability of Being a Girl?

To solve this, we need to look at the sample space for a two child family and apply the formula for conditional probability.
1. Define the Sample Space  
If we list the possible combinations for two children (where the order represents the first child and the second child), there are four equally likely outcomes:  
$(B, B)$ - Boy, Boy  
$(B, G)$ - Boy, Girl  
$(G, B)$ - Girl, Boy  
$(G, G)$ - Girl, Girl  
Each of these four outcomes has a probability of $1/4$.
2. Define the Events  
Event $A$ (What we want to find): Both children are girls.  
Outcomes for $A$: $\{(G, G)\}$  
Event $B$ (The given condition): At least one child is a girl.  
Outcomes for $B$: $\{(B, G), (G, B), (G, G)\}$  
3. Calculate the Conditional Probability  
We are looking for the probability of $A$ given that $B$ has already occurred, written as $P(A|B)$. The formula for conditional probability is:  
$$P(A|B) = \frac{P(A \cap B)}{P(B)}$$  
$P(B)$: There are $3$ favorable outcomes for event $B$ out of $4$ total possibilities, so $P(B) = 3/4$.  
$P(A \cap B)$: The probability that both events happen (both are girls AND at least one is a girl) is simply the probability of both being girls. There is $1$ favorable outcome for this, so $P(A \cap B) = 1/4$.  
Now, plug these into the formula:  
$$P(A|B) = \frac{1/4}{3/4} = \frac{1}{3}$$  
Therefore, the probability that both children are girls, given that at least one is a girl, is $1/3$.

# Q9) Information Available in a Discrete Random Variable

A factory inspects a product and records its quality using two different coding systems.

Each product falls into exactly one of the following four categories:

* $G_1$: good quality, produced on Machine 1
* $G_2$: good quality, produced on Machine 2

Assume the sample space is

$$
\Omega=\{G_1,G_2,D_1,D_2\},
$$

with probabilities

P(\{G_1\})=0.50,\qquad P(\{G_2\})=0.20,\qquad P(\{D_1\})=0.10,\qquad P(\{D_2\})=0.20.

Define two discrete random variables on the same probability space:

* $X$ records only whether the item is good or defective:
  $$
  X(G_1)=0,\quad X(G_2)=0,\quad X(D_1)=1,\quad X(D_2)=1.

Answer the following.

   $$

12. Find the marginal, joint, and conditional Shannon Entropy of $X$ and $Y$.

1. The Probability Space  
A probability space consists of three elements: a sample space ($\Omega$), a $\sigma$-algebra ($\mathcal{F}$), and a probability measure ($P$).  
Sample Space ($\Omega$): The set of all possible outcomes.  
$$\Omega = \{G_1, G_2, D_1, D_2\}$$  
$\sigma$-algebra ($\mathcal{F}$): Since the sample space is finite, $\mathcal{F}$ is the power set of $\Omega$ (denoted as $\mathcal{P}(\Omega)$). It contains all $2^4 = 16$ possible subsets of $\Omega$:  
$$\mathcal{F} = \{\emptyset, \{G_1\}, \{G_2\}, \{D_1\}, \{D_2\}, \{G_1, G_2\}, \dots, \Omega\}$$  
Probability Measure ($P$): A function mapping each event in $\mathcal{F}$ to a probability between 0 and 1. It is uniquely defined by the probabilities of the elementary events:  
$$P(\{G_1\}) = 0.50, \quad P(\{G_2\}) = 0.20, \quad P(\{D_1\}) = 0.10, \quad P(\{D_2\}) = 0.20$$  
2. $\sigma$-algebra generated by $X$, $\sigma(X)$  
The random variable $X$ records only whether the item is good or defective:$$X(G_1) = 0, \quad X(G_2) = 0, \quad X(D_1) = 1, \quad X(D_2) = 1.$$Thus the inverse images of the values of $X$ are$$X^{-1}(0) = \{G_1, G_2\},$$and$$X^{-1}(1) = \{D_1, D_2\}.$$Hence the $\sigma$-algebra generated by $X$ is$$\sigma(X) = \{\emptyset, \Omega, \{G_1, G_2\}, \{D_1, D_2\}\}.$$  
3. $\sigma$-algebra generated by $Y$, $\sigma(Y)$  
The random variable $Y$ records both quality and machine:$$Y(G_1) = 1, \quad Y(G_2) = 2, \quad Y(D_1) = 3, \quad Y(D_2) = 4.$$Since each outcome is assigned a distinct value, $Y$ distinguishes every point of $\Omega$. Therefore$$\sigma(Y) = \mathfrak{P}(\Omega).$$Explicitly,$$\sigma(Y) = \{\emptyset, \{G_1\}, \{G_2\}, \{D_1\}, \{D_2\}, \{G_1, G_2\}, \{G_1, D_1\}, \dots, \Omega\}.$$  
4. Show $\sigma(X) \subseteq \sigma(Y)$ and explain the information difference  
Every set in $\sigma(X)$ is also present in $\sigma(Y)$ (which contains all 16 possible subsets). Therefore:$$\sigma(X) \subseteq \sigma(Y)$$Explanation: $Y$ contains more information than $X$ because its $\sigma$-algebra is finer. Random variable $Y$ partitions the sample space into individual elements, giving complete information about both quality and the machine. Random variable $X$ only partitions the space into two larger blocks (Good vs. Defective), destroying the information about which machine was used.  
5. Marginal Distribution of $X$  
The probability distribution of $X$ by itself:$$P(X=0) = P(\{G_1, G_2\}) = 0.50 + 0.20 = 0.70$$$$P(X=1) = P(\{D_1, D_2\}) = 0.10 + 0.20 = 0.30$$  
6. Marginal Distribution of $Y$  
The probability distribution of $Y$ by itself:$$P(Y=1) = P(\{G_1\}) = 0.50$$$$P(Y=2) = P(\{G_2\}) = 0.20$$$$P(Y=3) = P(\{D_1\}) = 0.10$$$$P(Y=4) = P(\{D_2\}) = 0.20$$  
7. Joint Probabilities  
The joint probability $P(X=x, Y=y)$ is the probability of the intersection of their pre-images. Because $Y$ completely determines $X$, the joint probabilities are simply the probabilities of $Y$ when $X$ matches $Y$'s underlying quality, and 0 otherwise.$$P(X=0, Y=1) = P(\{G_1, G_2\} \cap \{G_1\}) = P(\{G_1\}) = 0.50$$$$P(X=0, Y=2) = P(\{G_1, G_2\} \cap \{G_2\}) = P(\{G_2\}) = 0.20$$$$P(X=1, Y=3) = P(\{D_1, D_2\} \cap \{D_1\}) = P(\{D_1\}) = 0.10$$$$P(X=1, Y=4) = P(\{D_1, D_2\} \cap \{D_2\}) = P(\{D_2\}) = 0.20$$  
8. Conditional Probabilities  
Using the definition $P(A \mid B) = \frac{P(A \cap B)}{P(B)}$:  
The conditional probabilities are$$P(X = 1 \mid Y = 3) = 1,$$because $Y = 3$ means the outcome is certainly $D_1$, which is defective.Similarly,$$P(X = 1 \mid Y = 4) = 1.$$Now$$P(Y = 3 \mid X = 1) = \frac{P(Y = 3, X = 1)}{P(X = 1)} = \frac{0.10}{0.30} = \frac{1}{3},$$and$$P(Y = 4 \mid X = 1) = \frac{P(Y = 4, X = 1)}{P(X = 1)} = \frac{0.20}{0.30} = \frac{2}{3}.$$  
9. Interpret the Difference in Conditioning  
Conditioning on $Y$ gives more precise information than conditioning on $X$. If we know $Y = 3$, then we know with certainty that the item is defective, so$$P(X = 1 \mid Y = 3) = 1.$$But if we know only that $X = 1$, we know the item is defective but we do not know which machine produced it. Then there is still uncertainty:$$P(Y = 3 \mid X = 1) = \frac{1}{3}, \quad P(Y = 4 \mid X = 1) = \frac{2}{3}.$$  
10. Information in $\sigma(X)$ vs $\sigma(Y)$  
The $\sigma$-algebra $\sigma(X)$ contains only the information about quality class:  
whether the item is good,  
or whether it is defective.  
It does not distinguish between Machine 1 and Machine 2.  
By contrast, $\sigma(Y)$ contains full information about the outcome. It tells us both whether the item is good or defective and which machine produced it. Thus $\sigma(Y)$ is strictly finer than $\sigma(X)$.  
A concise conclusion is:The two random variables are defined on the same sample space, but they generate different $\sigma$-algebras. Since$$\sigma(X) \subseteq \sigma(Y),$$the variable $Y$ carries more information than $X$. This is reflected in the conditional probabilities: conditioning on $Y$ can remove uncertainty that remains when conditioning only on $X$.

1. Probability Space $(\Omega, \mathfrak{F}, P)$  
The underlying probability space is defined as:  
Sample Space: $\Omega = \mathbb{R}^2$, where an element $\omega = (t, p)$ represents standardized temperature and pressure deviations.  
$\sigma$-algebra: $\mathfrak{F} = \mathfrak{B}(\mathbb{R}^2)$, which is the Borel $\sigma$-algebra on the plane.  
Probability Measure: $P$ is induced by the joint density function:$$f_{T,P}(t,p) = \frac{1}{4\pi}\exp\left( -\frac{t^2}{2} - \frac{p^2}{8} \right), \quad (t,p) \in \mathbb{R}^2$$For any Borel set $A \subseteq \mathbb{R}^2$, $P(A) = \iint_{A} f_{T,P}(t,p) dt dp$.  
2. $\sigma$-algebra generated by $X$  
The measurement $X$ is defined by $X(t,p) = t$, recording only temperature.Since $X^{-1}(B) = \{(t,p) \in \mathbb{R}^2 : t \in B\} = B \times \mathbb{R}$, we obtain:$$\sigma(X) = \{ B \times \mathbb{R} : B \in \mathfrak{B}(\mathbb{R}) \}$$This $\sigma$-algebra consists of "vertical strips" that depend solely on the temperature coordinate.  
3. $\sigma$-algebra generated by $Y$  
$Y$ is the identity map $Y(t,p) = (t,p)$, recording the full system state. Its generated $\sigma$-algebra is:$$\sigma(Y) = \mathfrak{B}(\mathbb{R}^2)$$Because $Y^{-1}(A) = A$ for every Borel set, $\sigma(Y)$ can distinguish any Borel subset of the plane.  
4. Information Inclusion $\sigma(X) \subseteq \sigma(Y)$  
Every set of the form $B \times \mathbb{R}$ is a Borel subset of $\mathbb{R}^2$, thus $\sigma(X) \subseteq \sigma(Y)$.
The inclusion is strict because sets such as $[-1, 1] \times [-2, 2]$ belong to $\sigma(Y)$ but not to $\sigma(X)$ as they depend on both coordinates. This proves $Y$ contains strictly more information than $X$.
5. Marginal Density of $X$  
By integrating out the pressure component $p$:$$f_X(x) = \int_{-\infty}^{\infty} \frac{1}{4\pi}\exp\left( -\frac{x^2}{2} - \frac{p^2}{8} \right) dp = \frac{1}{\sqrt{2\pi}} \exp\left( -\frac{x^2}{2} \right)$$Thus, $X \sim N(0, 1)$.  
6. Marginal Density of $Y$  
Since $Y = (T, P)$, its marginal density is the joint density:$$f_Y(t,p) = \frac{1}{4\pi}\exp\left( -\frac{t^2}{2} - \frac{p^2}{8} \right)$$$Y$ is a bivariate normal random vector with independent components.  
7. Computed Probabilities  
$P(X \leq 0)$: By symmetry of the standard normal, $P(T \leq 0) = \mathbf{1/2}$.$P(X > 1)$: $1 - \Phi(1) \approx \mathbf{0.1587}$.  
$P(Y \in (-\infty, 0] \times \mathbb{R})$: Equivalent to $P(T \leq 0) = \mathbf{1/2}$.$P(Y \in [-1, 1] \times [-2, 2])$: Since $T, P$ are independent:$$P(-1 \leq T \leq 1)P(-2 \leq P \leq 2) = (2\Phi(1) - 1)^2 \approx \mathbf{0.466}$$  
8. Compute the conditional distribution of $X$ given $Y = (t, p)$  
Since $Y = (T, P)$ represents the full measurement of both temperature and pressure, it completely determines the value of $X$ ($T$). Therefore, the conditional distribution of $X$ given a specific observation $Y = (t, p)$ is degenerate at the value $t$:$$P(X = t \mid Y = (t,p)) = 1.$$Equivalently, the conditional law of $X$ given $Y = (t,p)$ is expressed as:$$\delta_t,$$which is the Dirac mass at $t$. This means that once both sensors are read, the temperature deviation is known with absolute certainty.  
9. Compute the conditional distribution of $Y$ given $X = t$  
The conditional distribution of $Y$ given $X = t$ represents the behavior of the random vector $(t, P)$ under the condition that temperature is fixed. Since $T$ and $P$ are independent, the pressure $P$ remains normally distributed with its original variance ($P \sim N(0,4)$). Thus:$$Y \mid X = t \sim (t, P).$$Formally, the conditional probability measure $P_{Y|X=t}$ is the product measure:$$\delta_t \otimes \mathcal{N}(0, 4).$$The conditional density $f_{Y|X}((u, v) \mid t)$ can be expressed as:$$f_{Y|X}((u, v) \mid t) = \delta_t(u) \cdot \frac{1}{\sqrt{8\pi}}\exp\left( -\frac{v^2}{8} \right).$$Informally: Given that $X = t$, the temperature component is fixed at $t$ (represented by the Dirac mass $\delta_t$), while the pressure component remains normally distributed.  
10. Physical Interpretation of Conditioning  
Conditioning on $X$: We only know the temperature; pressure remains random and uncertain.  
Conditioning on $Y$: We know both temperature and pressure; all coordinates are known exactly, removing all uncertainty.  
11. Measurability and Projections  
Define the projection $\pi_1(t,p) = t$. Then $X = \pi_1 \circ Y$.
Since $\pi_1$ is Borel measurable and $X$ is a measurable function of $Y$, it follows that $\sigma(X) \subseteq \sigma(Y)$, providing a formal explanation for why $Y$ is more informative.

13. Compute the differential entropy of $X$  
The random variable $X$ represents the temperature deviation and follows a standard normal distribution, $X \sim N(0,1)$. The differential entropy for a normal distribution $N(\mu, \sigma^2)$ is given by the formula $h(X) = \frac{1}{2}\ln(2\pi e \sigma^2)$.Substituting $\sigma^2 = 1$:$$h(X) = \frac{1}{2}\ln(2\pi e)$$Numerical Value:$$h(X) \approx 1.4189 \text{ nats}$$  
14. Compute the differential entropy of $Y$  
The random variable $Y = (T, P)$ represents the full sensor state, where $T$ and $P$ are independent normal components with variances $\sigma_T^2 = 1$ and $\sigma_P^2 = 4$. For independent variables, the joint differential entropy is the sum of their individual entropies:$$h(Y) = h(T) + h(P)$$Using the entropy formula for both:  
$h(T) = \frac{1}{2}\ln(2\pi e)$  
$h(P) = \frac{1}{2}\ln(2\pi e \cdot 4) = \frac{1}{2}\ln(8\pi e)$  
  
  Calculation:$$h(Y) = \frac{1}{2}\ln(2\pi e) + \frac{1}{2}\ln(8\pi e) = \ln(2\pi e) + \ln 2$$Numerical Value:$$h(Y) \approx 3.5310 \text{ nats}$$  
15. Compute the conditional differential entropy $h(Y|X)$  
The conditional differential entropy $h(Y|X)$ measures the uncertainty remaining in the full sensor state $Y$ once the temperature $X$ is known. Since $T$ and $P$ are independent, knowing $T$ provides no information about $P$. Therefore, the remaining uncertainty in $Y$ is exactly the uncertainty in $P$:$$h(Y|X) = h(P)$$Calculation:$$h(Y|X) = \frac{1}{2}\ln(2\pi e \cdot 4) = \frac{1}{2}\ln(8\pi e)$$Numerical Value:$$h(Y|X) \approx 2.1121 \text{ nats}$$Physical Interpretation:Once the temperature deviation is measured and known, the only remaining uncertainty in the system is the pressure deviation.  
16. The $h(X \mid Y)$ Discontinuity  
While in discrete cases $H(X \mid Y) = 0$ if $Y$ determines $X$, in the continuous setting, $h(X \mid Y)$ formally tends to $-\infty$. This reflects that exact deterministic dependence in continuous space corresponds to a singular distribution, highlighting a fundamental difference between discrete and differential entropy.
